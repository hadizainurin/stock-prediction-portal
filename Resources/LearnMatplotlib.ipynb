## Bar Chart

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [1000, 1500, 1200, 1700, 2000, 2500]
```

```python
plt.figure(figsize=(10,6)) # To change chart size
plt.bar(months, revenue)
plt.title('Monthly Revenue Bar Chart')
plt.xlabel('Months')
plt.ylabel('Revenue')
plt.show()
```

#### Subplots and Legends

```python
days = [1, 2, 3, 4, 5]
original_price = [10, 20, 25, 30, 40]
predicted_price = [8, 21, 25, 28, 38]
```

```python
plt.plot(days, original_price, 'b-o',label='Original Price')
plt.plot(days, predicted_price, 'r-o',label='Predicted Price')
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()
plt.show()
```

## Subplots

```python
# First subplot - Original Price
plt. subplot(3, 1, 1) # 3 rows, 1 column, 1st plot
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.title('Original Price over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()

# Second subplot - Predicted Price
plt. subplot(3, 1, 2) # 3 rows, 1 column, 2nd plot
plt.plot(days, predicted_price, 'r-o', label='Predicted Price')
plt.title('Predicted Price over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()

# Third subplot - Original vs Predicted Price
price_difference = [o - p for o, p in zip(original_price, predicted_price)]
plt.subplot(3, 1, 3) # 3 rows, 1 column, 3rd plot
plt.bar(days, price_difference, color='purple', label='Price Difference')
plt.title('Original vs Predicted Price')
plt.xlabel('Days')
plt.ylabel('Price difference')
plt.legend()

plt.tight_layout()
plt.savefig('example_prediction.png', dpi=300) # Save the image
```
