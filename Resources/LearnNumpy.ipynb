#### Appending and insercting

```python
zeroes
```

```python
# Create a new array
new_array = np.ones((3, 2, 3))
new_array
```

```python
appd = np.append(zeroes, new_array, axis=0) # axis 0 is row, axis 1 is column, axis 2 is layer
appd
```

#### Insert

```python
new_array
```

```python
insterd_array = np.insert(zeroes, 2, new_array, axis=0) # Insert new_array into zeroes at index 2. axis 0 is row, axis 1 is column, axis 2 is layer
insterd_array
```
